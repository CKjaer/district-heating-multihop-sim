```python
# TEG specs from Koolivand et al. 2025
COUPLER_COLD_TEMP = 38.5 # C
COUPLER_WARM_TEMP = 53.2 # C
SENSITIVITY = 36.5e-3 # V/C
INTERNAL_RESISTANCE = 2.17 # Ohm

# Super capacitor specs
CAPACITANCE = 1.0 # F 
REGULATOR_THRESHOLD = 3.5 # V
MAX_VOLTAGE = 5.3 # V

delta_temp = COUPLER_WARM_TEMP - COUPLER_COLD_TEMP
V_oc = delta_temp * SENSITIVITY
print(f"Open circuit voltage: {V_oc:.3f} V")

# Power output from maximum power transfer theorem
# Assuming load resistance is equal to internal resistance
P_max = V_oc**2 / (4 * INTERNAL_RESISTANCE)
print(f"Maximum power output: {P_max*1e3:.3f} mW")

# Available energy in super capacitor
E_buffer = 0.5 * CAPACITANCE * (MAX_VOLTAGE**2 - REGULATOR_THRESHOLD**2)
print(f"Available energy: {E_buffer:.3f} J")

# Average harvested power (takes 4 hours to charge)
P_avg = E_buffer / (4 * 60 * 60)
print(f"Average harvested power: {P_avg*1e3:.3f} mW") # Reduced maybe due to regulator start-up
```

## Temperature Loss in Pipeline

Energy balance equation for steady-state fluid

$$
q - w = h_2 - h_1
$$

Where $\Delta h$ is enthalpy change for the fluid. From the definition of ethalpy
$$
\Delta h = c(T_2 - T_1) + v(P_2-P_1)
$$

$c$ specific temperature 

$P$ pressure 

$v$ velocity

For an ideal gas $\Delta h\approx c(T_2-T_1)$ and no work $w=0$


$$

\Delta T = \frac{q'L}{\dot{m}c}
$$


```python
import numpy as np
c = 4184 # J/(kg C)
q = 8.2 # W/m
L = 1e3 # m
cubic_flow = 1 # m3/h
water_density = 1000 # kg/m3
m = (cubic_flow * water_density) / 3600 # kg/s

Delta_T = lambda L : (q * L) / (m * c)
print(f"Temperature loss is {Delta_T(L):.2f} C for {L} m pipe")

qs = m * c * np.abs(80 - 77.1) / 500
qr = m * c * np.abs(40 - 40.3) / 500
```
